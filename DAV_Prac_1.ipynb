STEP 1 : IMPORT THE PANDAS LIBRARY

```python
# import pandas
import pandas as pd
```

STEP 2 : CREATE SAMPLE DATA

```python
# create a student dataset
data = {
     "ID": [101, 102, 103, 104, 105, 106],
     "Name": ["Himay", "Janvi", "bhavya", "kavy", "dharmik", "sneh"],
     "Age": [20, 22, 24, 25, 30, 32],
     "City": ["surat","navsari","valsad","vapi","gandhinagar","Delhi"],
     "Marks": [85, 92, 78, 88, 84, 90]
       }
```

STEP 3: CREATE A DATAFRAME

```python
# Create a DataFrame from the given data and dispaly the data
df = pd.DataFrame(data)

df

```

STEP 4 : SAVE DATA AS A CSV FILE

```python
# save data in "students.csv" file
df.to_csv("students.csv", index=False)

print("csv file created successfully.")
```

STEP 5 : SAVE DATA AS A EXCEL FILE

```python
# save data in "students.xlsx" excel file
df.to_excel("students.xlsx", index=False)

print("Excel file created successfully.")
```

STEP 6 : SAVE DATA AS JASON FILE

```python
# save data in "students.json" json file
df.to_json("students.json", orient="records", index=4)

print("json file created successfully.")
```

STEP 7 : READ THE CSV FILE
